```python
# ============================================================
# DATA STORYTELLING FOR INTERACTIVE DECISION ANALYTICS
# DATA PROCESSING + EDA + FEATURE ENGINEERING
# ============================================================

import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns

# ============================================================
# 1. LOAD THE FOUR DATASETS
# ============================================================

supermarket = pd.read_csv(
    r"C:\Users\rathn\DATA_STORY_TELLING\datasets\supermarket_dataset.csv"
)

global_superstore = pd.read_csv(
    r"C:\Users\rathn\DATA_STORY_TELLING\datasets\globalsuper.csv"
)

superstore = pd.read_csv(
    r"C:\Users\rathn\DATA_STORY_TELLING\datasets\Superstore Dataset.csv"
)

online_sales = pd.read_csv(
    r"C:\Users\rathn\DATA_STORY_TELLING\datasets\onlinesales.csv"
)

# Store all datasets together
datasets = {
    "Supermarket": supermarket,
    "Global Superstore": global_superstore,
    "Superstore": superstore,
    "Online Sales": online_sales
}

# ============================================================
# 2. DATASET SIZE
# ============================================================

print("DATASET SIZES")
print("=" * 60)

for name, df in datasets.items():
    print(f"{name}: {df.shape}")

# ============================================================
# 3. VIEW FIRST 5 ROWS
# ============================================================

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)
    display(df.head())

# ============================================================
# 4. COLUMN NAMES
# ============================================================

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name, "- COLUMNS")
    print("=" * 60)

    for col in df.columns:
        print(repr(col))

# ============================================================
# 5. DATA TYPES
# ============================================================

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name, "- DATA TYPES")
    print("=" * 60)

    print(df.dtypes)

# ============================================================
# 6. MISSING VALUES
# ============================================================

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name, "- MISSING VALUES")
    print("=" * 60)

    print(df.isnull().sum())

# ============================================================
# 7. MISSING VALUE PERCENTAGE
# ============================================================

for name, df in datasets.items():

    print("\n" + "=" * 60)
    print(name, "- MISSING VALUE PERCENTAGE")
    print("=" * 60)

    missing_percent = (
        df.isnull().sum() / len(df) * 100
    )

    print(
        missing_percent
        .sort_values(ascending=False)
    )

# ============================================================
# 8. DUPLICATE CHECK
# ============================================================

print("\nDUPLICATE RECORDS")
print("=" * 60)

for name, df in datasets.items():
    print(
        name,
        "duplicates:",
        df.duplicated().sum()
    )

# ============================================================
# 9. REMOVE DUPLICATES
# ============================================================

print("\nREMOVING DUPLICATES")
print("=" * 60)

for name, df in datasets.items():

    before = df.duplicated().sum()

    df.drop_duplicates(
        inplace=True
    )

    after = df.duplicated().sum()

    print(
        f"{name}: Before = {before}, After = {after}"
    )

# ============================================================
# 10. STANDARDIZE COLUMN NAMES
# ============================================================

for name, df in datasets.items():

    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_")
        .str.replace("-", "_")
    )

print("\nColumn names standardized.")

# ============================================================
# 11. FIND DATE COLUMNS
# ============================================================

print("\nDATE COLUMNS")
print("=" * 60)

for name, df in datasets.items():

    date_columns = [
        col
        for col in df.columns
        if "date" in col.lower()
    ]

    print(name, ":", date_columns)

# ============================================================
# 12. CONVERT DATE COLUMNS
# ============================================================

for name, df in datasets.items():

    date_columns = [
        col
        for col in df.columns
        if "date" in col.lower()
    ]

    for col in date_columns:

        df[col] = pd.to_datetime(
            df[col],
            errors="coerce"
        )

print("\nDate conversion completed.")

# ============================================================
# 13. NUMERICAL COLUMNS
# ============================================================

for name, df in datasets.items():

    print("\n" + "=" * 60)
    print(name, "- NUMERICAL COLUMNS")
    print("=" * 60)

    print(
        df.select_dtypes(
            include=np.number
        ).columns.tolist()
    )

# ============================================================
# 14. CATEGORICAL COLUMNS
# ============================================================

for name, df in datasets.items():

    print("\n" + "=" * 60)
    print(name, "- CATEGORICAL COLUMNS")
    print("=" * 60)

    print(
        df.select_dtypes(
            include=["object", "string"]
        ).columns.tolist()
    )

# ============================================================
# 15. BASIC STATISTICS
# ============================================================

for name, df in datasets.items():

    print("\n" + "=" * 60)
    print(name, "- BASIC STATISTICS")
    print("=" * 60)

    display(df.describe(include="all").T)

# ============================================================
# 16. EDA - NUMERICAL DISTRIBUTIONS
# ============================================================

for name, df in datasets.items():

    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns

    for col in numerical_columns[:5]:

        plt.figure(figsize=(8, 5))
        plt.hist(df[col].dropna(), bins=30)
        plt.title(f"{name} - {col} Distribution")
        plt.xlabel(col)
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()

# ============================================================
# 17. EDA - CATEGORICAL ANALYSIS
# ============================================================

for name, df in datasets.items():

    categorical_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for col in categorical_columns[:3]:

        value_counts = df[col].value_counts().head(10)

        plt.figure(figsize=(8, 5))
        value_counts.plot(kind="bar")
        plt.title(f"{name} - {col}")
        plt.xlabel(col)
        plt.ylabel("Count")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

# ============================================================
# 18. CORRELATION ANALYSIS
# ============================================================

for name, df in datasets.items():

    numerical_df = df.select_dtypes(
        include=np.number
    )

    if numerical_df.shape[1] >= 2:

        plt.figure(figsize=(10, 7))

        sns.heatmap(
            numerical_df.corr(),
            annot=True,
            cmap="coolwarm",
            fmt=".2f"
        )

        plt.title(f"{name} - Correlation Analysis")
        plt.tight_layout()
        plt.show()

# ============================================================
# 19. FINAL DATA QUALITY CHECK
# ============================================================

print("\nFINAL DATA QUALITY CHECK")
print("=" * 60)

for name, df in datasets.items():

    print("\n", name)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Duplicates:", df.duplicated().sum())
    print("Missing Values:", df.isnull().sum().sum())

# ============================================================
## ============================================================
# 20. FEATURE ENGINEERING
# ============================================================

print("\nFEATURE ENGINEERING")
print("=" * 60)

for name, df in datasets.items():

    print("\n", name)

    # --------------------------------------------------------
    # Create Year, Month and Quarter from Date Columns
    # --------------------------------------------------------

    date_columns = [
        col for col in df.columns
        if "date" in col.lower()
    ]

    for col in date_columns:

        df[col] = pd.to_datetime(
            df[col],
            errors="coerce"
        )

        df[col + "_Year"] = df[col].dt.year
        df[col + "_Month"] = df[col].dt.month
        df[col + "_Quarter"] = df[col].dt.quarter

    # --------------------------------------------------------
    # Convert Sales and Profit to Numeric
    # --------------------------------------------------------

    for col in df.columns:

        if col.lower() in ["sales", "profit", "quantity", "discount"]:

            df[col] = pd.to_numeric(
                df[col],
                errors="coerce"
            )

    # --------------------------------------------------------
    # Create Profit Margin
    # --------------------------------------------------------

    if "Sales" in df.columns and "Profit" in df.columns:

        df["Profit_Margin"] = np.where(
            df["Sales"] != 0,
            (df["Profit"] / df["Sales"]) * 100,
            0
        )

        print("Profit Margin created.")

    else:
        print("Sales/Profit not available.")

    print("Feature engineering completed.")
# 21. DISPLAY CREATED FEATURES
# ============================================================

print("\nFINAL FEATURES")
print("=" * 60)

for name, df in datasets.items():

    print("\n" + name)
    print(df.columns.tolist())

# ============================================================
# 22. SAVE PROCESSED DATASETS
# ============================================================

output_folder = (
    r"C:\Users\rathn\DATA_STORY_TELLING\processed_data"
)

os.makedirs(
    output_folder,
    exist_ok=True
)

supermarket.to_csv(
    os.path.join(
        output_folder,
        "supermarket_cleaned.csv"
    ),
    index=False
)

global_superstore.to_csv(
    os.path.join(
        output_folder,
        "global_superstore_cleaned.csv"
    ),
    index=False
)

superstore.to_csv(
    os.path.join(
        output_folder,
        "superstore_cleaned.csv"
    ),
    index=False
)

online_sales.to_csv(
    os.path.join(
        output_folder,
        "online_sales_cleaned.csv"
    ),
    index=False
)

print("\n" + "=" * 60)
print("DATA PROCESSING + EDA + FEATURE ENGINEERING COMPLETED")
print("=" * 60)

print("Cleaned datasets saved at:")
print(output_folder)


```

```python
# ============================================================
# PREPROCESSING + EDA FOR THE NEXT 3 DATASETS
# Global Superstore
# Superstore
# Online Sales
# ============================================================

# Select only the next 3 datasets
next_datasets = {
    "Global Superstore": global_superstore,
    "Superstore": superstore,
    "Online Sales": online_sales
}


# ============================================================
# 1. DATASET INSPECTION
# ============================================================

for name, df in next_datasets.items():

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print("Shape:", df.shape)

    print("\nFirst 5 Rows:")
    display(df.head())

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)


# ============================================================
# 2. MISSING VALUE ANALYSIS
# ============================================================

for name, df in next_datasets.items():

    print("\n" + "=" * 70)
    print(name, "- MISSING VALUES")
    print("=" * 70)

    missing = df.isnull().sum()

    missing_percent = (
        df.isnull().sum() / len(df) * 100
    )

    missing_table = pd.DataFrame({
        "Missing Values": missing,
        "Missing Percentage": missing_percent
    })

    display(
        missing_table[
            missing_table["Missing Values"] > 0
        ].sort_values(
            "Missing Percentage",
            ascending=False
        )
    )


# ============================================================
# 3. DUPLICATE ANALYSIS
# ============================================================

for name, df in next_datasets.items():

    print(
        name,
        "Duplicate Rows:",
        df.duplicated().sum()
    )


# ============================================================
# 4. REMOVE DUPLICATES
# ============================================================

for name, df in next_datasets.items():

    before = df.duplicated().sum()

    df.drop_duplicates(
        inplace=True
    )

    after = df.duplicated().sum()

    print(
        f"{name}: Before = {before}, After = {after}"
    )


# ============================================================
# 5. STANDARDIZE COLUMN NAMES
# ============================================================

for name, df in next_datasets.items():

    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_")
        .str.replace("-", "_")
    )

print("\nColumn names standardized.")


# ============================================================
# 6. HANDLE MISSING VALUES
# ============================================================

for name, df in next_datasets.items():

    # Numerical columns
    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns

    for col in numerical_columns:

        if df[col].isnull().sum() > 0:

            df[col] = df[col].fillna(
                df[col].median()
            )

    # Categorical columns
    categorical_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for col in categorical_columns:

        if df[col].isnull().sum() > 0:

            mode_value = df[col].mode()

            if len(mode_value) > 0:

                df[col] = df[col].fillna(
                    mode_value[0]
                )

print("\nMissing values handled.")


# ============================================================
# 7. AUTOMATIC DATE COLUMN DETECTION
# ============================================================

for name, df in next_datasets.items():

    date_columns = [
        col
        for col in df.columns
        if "date" in col.lower()
    ]

    print(
        f"{name} Date Columns:",
        date_columns
    )

    for col in date_columns:

        df[col] = pd.to_datetime(
            df[col],
            errors="coerce"
        )


# ============================================================
# 8. NUMERICAL AND CATEGORICAL FEATURES
# ============================================================

for name, df in next_datasets.items():

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns.tolist()

    print("Numerical Columns:")
    print(numerical_columns)

    print("\nCategorical Columns:")
    print(categorical_columns)


# ============================================================
# 9. FINAL PREPROCESSING VALIDATION
# ============================================================

print("\nFINAL PREPROCESSING CHECK")
print("=" * 70)

for name, df in next_datasets.items():

    print("\n", name)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Missing Values:", df.isnull().sum().sum())
    print("Duplicate Rows:", df.duplicated().sum())


# ============================================================
# 10. DESCRIPTIVE STATISTICS
# ============================================================

for name, df in next_datasets.items():

    print("\n" + "=" * 70)
    print(name, "- DESCRIPTIVE STATISTICS")
    print("=" * 70)

    display(
        df.describe(
            include="all"
        ).T
    )


# ============================================================
# 11. EDA - UNIVARIATE ANALYSIS
# ============================================================

for name, df in next_datasets.items():

    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns.tolist()

    for col in numerical_columns:

        plt.figure(figsize=(8, 5))

        plt.hist(
            df[col].dropna(),
            bins=30
        )

        plt.title(
            f"{name} - {col} Distribution"
        )

        plt.xlabel(col)
        plt.ylabel("Frequency")

        plt.tight_layout()
        plt.show()


# ============================================================
# 12. EDA - CATEGORICAL ANALYSIS
# ============================================================

for name, df in next_datasets.items():

    categorical_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns.tolist()

    for col in categorical_columns[:5]:

        counts = (
            df[col]
            .value_counts()
            .head(10)
        )

        plt.figure(figsize=(9, 5))

        counts.plot(
            kind="bar"
        )

        plt.title(
            f"{name} - {col}"
        )

        plt.xlabel(col)
        plt.ylabel("Count")

        plt.xticks(
            rotation=45,
            ha="right"
        )

        plt.tight_layout()
        plt.show()


# ============================================================
# 13. EDA - OUTLIER ANALYSIS
# ============================================================

for name, df in next_datasets.items():

    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns.tolist()

    for col in numerical_columns:

        plt.figure(figsize=(8, 4))

        plt.boxplot(
            df[col].dropna()
        )

        plt.title(
            f"{name} - {col} Boxplot"
        )

        plt.ylabel(col)

        plt.tight_layout()
        plt.show()


# ============================================================
# 14. EDA - CORRELATION ANALYSIS
# ============================================================

for name, df in next_datasets.items():

    numerical_df = df.select_dtypes(
        include=np.number
    )

    if numerical_df.shape[1] >= 2:

        plt.figure(figsize=(10, 7))

        sns.heatmap(
            numerical_df.corr(),
            annot=True,
            fmt=".2f",
            cmap="coolwarm"
        )

        plt.title(
            f"{name} - Correlation Analysis"
        )

        plt.tight_layout()
        plt.show()


# ============================================================
# 15. EDA - TREND ANALYSIS
# ============================================================

for name, df in next_datasets.items():

    date_columns = [
        col
        for col in df.columns
        if "date" in col.lower()
    ]

    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns.tolist()

    if (
        len(date_columns) > 0
        and len(numerical_columns) > 0
    ):

        date_col = date_columns[0]
        value_col = numerical_columns[0]

        temp = df.dropna(
            subset=[date_col]
        ).copy()

        temp["Year_Month"] = (
            temp[date_col]
            .dt.to_period("M")
            .astype(str)
        )

        trend = (
            temp.groupby("Year_Month")[value_col]
            .sum()
            .reset_index()
        )

        plt.figure(figsize=(12, 5))

        plt.plot(
            trend["Year_Month"],
            trend[value_col],
            marker="o"
        )

        plt.title(
            f"{name} - Monthly Trend"
        )

        plt.xlabel("Month")
        plt.ylabel(value_col)

        plt.xticks(
            rotation=45
        )

        plt.tight_layout()
        plt.show()


# ============================================================
# 16. FEATURE ENGINEERING
# ============================================================

print("\nFEATURE ENGINEERING")
print("=" * 70)

for name, df in next_datasets.items():

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    # --------------------------------------------------------
    # Create Year, Month and Quarter from Date Columns
    # --------------------------------------------------------

    date_columns = [
        col for col in df.columns
        if "date" in col.lower()
    ]

    for col in date_columns:

        if pd.api.types.is_datetime64_any_dtype(df[col]):

            df[col + "_Year"] = df[col].dt.year
            df[col + "_Month"] = df[col].dt.month
            df[col + "_Quarter"] = df[col].dt.quarter

    # --------------------------------------------------------
    # Convert Important Numerical Columns
    # --------------------------------------------------------

    for col in df.columns:

        if col.lower() in [
            "sales",
            "profit",
            "quantity",
            "discount"
        ]:

            df[col] = pd.to_numeric(
                df[col],
                errors="coerce"
            )

    # --------------------------------------------------------
    # Create Profit Margin
    # --------------------------------------------------------

    sales_col = None
    profit_col = None

    for col in df.columns:

        if col.lower() == "sales":
            sales_col = col

        if col.lower() == "profit":
            profit_col = col

    if sales_col and profit_col:

        df["Profit_Margin"] = np.where(
            df[sales_col] != 0,
            (df[profit_col] / df[sales_col]) * 100,
            0
        )

        print("Created: Profit_Margin")

    # --------------------------------------------------------
    # Create Sales Per Quantity
    # --------------------------------------------------------

    quantity_col = None

    for col in df.columns:

        if col.lower() == "quantity":
            quantity_col = col

    if sales_col and quantity_col:

        df["Sales_Per_Quantity"] = np.where(
            df[quantity_col] != 0,
            df[sales_col] / df[quantity_col],
            0
        )

        print("Created: Sales_Per_Quantity")

    print("Feature engineering completed.")


# ============================================================
# 17. CHECK NEW FEATURES
# ============================================================

print("\nNEW FEATURES CREATED")
print("=" * 70)

for name, df in next_datasets.items():

    print("\n" + name)

    new_features = [
        col for col in df.columns
        if (
            "_Year" in col
            or "_Month" in col
            or "_Quarter" in col
            or col in [
                "Profit_Margin",
                "Sales_Per_Quantity"
            ]
        )
    ]

    print(new_features)


# ============================================================
# 18. SAVE THE THREE PROCESSED DATASETS
# ============================================================

output_folder = (
    r"C:\Users\rathn\DATA_STORY_TELLING\processed_data"
)

os.makedirs(
    output_folder,
    exist_ok=True
)

for name, df in next_datasets.items():

    filename = (
        name.lower()
        .replace(" ", "_")
        + "_preprocessed.csv"
    )

    df.to_csv(
        os.path.join(
            output_folder,
            filename
        ),
        index=False
    )

print("\n" + "=" * 70)
print("PREPROCESSING + EDA + FEATURE ENGINEERING COMPLETED")
print("=" * 70)

print("Datasets completed:")
print("1. Global Superstore")
print("2. Superstore")
print("3. Online Sales")

print("\nFiles saved in:")
print(output_folder)
```

```python
# ============================================================
# KPI ANALYSIS
# ============================================================

for name, df in selected_datasets.items():

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    # Sales
    sales_cols = [
        col for col in df.columns
        if col.lower() == "sales"
    ]

    if sales_cols:

        sales = pd.to_numeric(
            df[sales_cols[0]],
            errors="coerce"
        )

        print("Total Sales:", round(sales.sum(), 2))
        print("Average Sales:", round(sales.mean(), 2))

    # Profit
    profit_cols = [
        col for col in df.columns
        if col.lower() == "profit"
    ]

    if profit_cols:

        profit = pd.to_numeric(
            df[profit_cols[0]],
            errors="coerce"
        )

        print("Total Profit:", round(profit.sum(), 2))
        print("Average Profit:", round(profit.mean(), 2))

    # Quantity
    quantity_cols = [
        col for col in df.columns
        if col.lower() == "quantity"
    ]

    if quantity_cols:

        quantity = pd.to_numeric(
            df[quantity_cols[0]],
            errors="coerce"
        )

        print("Total Quantity:", round(quantity.sum(), 2))
        print("Average Quantity:", round(quantity.mean(), 2))
```

```python
# ============================================================
# DATA STORY 6 - SALES TREND
# ============================================================

for name, df in selected_datasets.items():

    date_columns = [
        col for col in df.columns
        if "date" in col.lower()
    ]

    if date_columns and "Sales" in df.columns:

        date_col = date_columns[0]

        temp = df.dropna(
            subset=[date_col]
        ).copy()

        temp["Year_Month"] = (
            temp[date_col]
            .dt.to_period("M")
            .astype(str)
        )

        trend = (
            temp.groupby("Year_Month")["Sales"]
            .sum()
            .reset_index()
        )

        print("\n" + "=" * 70)
        print(name)
        print("=" * 70)

        display(trend.head(20))
```
